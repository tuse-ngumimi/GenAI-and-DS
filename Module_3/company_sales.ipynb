```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'C:\Users\Ngumimi\genai-ds\Module_3\company_sales_data.csv', encoding='latin1')

sns.set_theme(style="whitegrid")
```

```python
# Below is the lineplot
plt.figure(figsize=(9, 5))
plt.plot(df['month_number'], df['total_profit'], marker='o', color="#2D39A3", linewidth=3)
plt.title('Total Profit per Month', fontsize=13, fontweight='bold', fontname= "Times New Roman")
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.xticks(df['month_number'])
plt.tight_layout()
plt.savefig('line_plot.png', dpi=150)
plt.show()
```

```python
# Below is the scatter plot 
plt.figure(figsize=(9, 5))
plt.scatter(df['total_units'], df['total_profit'], color="#570D5A", s=80, edgecolor='black')
plt.title('Total Units Sold vs Total Profit', fontsize=14, fontweight='bold', fontname="Times New Roman")
plt.xlabel('Total Units Sold')
plt.ylabel('Total Profit')
plt.tight_layout()
plt.savefig('scatter_plot.png', dpi=150)
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

products_list = ['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer'] 
fig, axes = plt.subplots(2, 3, figsize=(15, 8)) 

for ax, product in zip(axes.flatten(), products_list): 
    sns.histplot(df[product], kde=True, ax=ax, color="#DD3939", bins=6, kde_kws={'cut': 2}) 
    ax.set_title(product.capitalize(), fontsize=12, pad=10) 
    ax.set_xlabel('Units Sold', fontsize=10)
    ax.set_ylabel('Count', fontsize=10)

plt.suptitle('Distribution of Monthly Sales by Product', fontsize=16, fontweight='bold', y=0.98) 

plt.tight_layout(h_pad=3, w_pad=2) 
plt.savefig('histograms.png', dpi=150) 
plt.show()

```

```python
product_sales = df[products_list].melt(var_name='Product', value_name='Units Sold')

plt.figure(figsize=(11, 6))
sns.boxplot(data=product_sales, x='Product', y='Units Sold', hue='Product', palette='Set2', legend=False)
plt.title('Spread of Monthly Sales by Product', fontsize=14, fontweight='bold')
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig('box_plot.png', dpi=150)
plt.show()
```
